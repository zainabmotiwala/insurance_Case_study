# Forecasting Model Benchmark: Maharashtra Vehicle Registrations

Comparing Seasonal-Naive, Holt-Winters, SARIMA (with COVID exog), and Lagged Gradient Boosting.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load precomputed benchmark metrics and diagnostics
metrics_df = pd.read_csv('reports/forecast_metrics_summary.csv')
rolling_df = pd.read_csv('reports/rolling_origin_metrics.csv')
rolling_df['Date'] = pd.to_datetime(rolling_df['Date'])

print('Forecast Benchmark Metrics across Evaluated Powertrains:')
metrics_df
```

## 1. Out-of-Sample Performance Comparison

```python
total_metrics = metrics_df[metrics_df['Series']=='Total'].sort_values('Overall_WAPE_%')
print('Total Statewide Registrations Ranking:')
total_metrics[['Model', 'Val_MAE', 'Val_WAPE_%', 'Test_MAE', 'Test_WAPE_%', 'Overall_MAE', 'Overall_WAPE_%']]
```

## 2. Rolling-Origin Backtesting Walk-Forward Evaluation

```python
plt.figure(figsize=(12, 5))
plt.plot(rolling_df['Date'], rolling_df['Actual'], 'k-o', label='Actual Observed')
plt.plot(rolling_df['Date'], rolling_df['SNaive_Pred'], '--', color='gray', label='Seasonal-Naive')
plt.plot(rolling_df['Date'], rolling_df['SARIMA_Pred'], '-', color='blue', lw=2, label='SARIMA (COVID-exog)')
plt.title('Rolling-Origin 1-Step Ahead Out-of-Sample Predictions (2024–2026)', fontsize=12, fontweight='bold')
plt.ylabel('Monthly Registrations')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 3. Residual Diagnostics & Ljung-Box Autocorrelation

```python
from statsmodels.stats.diagnostic import acorr_ljungbox
# Load SARIMA residuals
diagnostics_df = pd.read_csv('reports/statewide_monthly_timeseries_diagnostics.csv')
lb_res = acorr_ljungbox(diagnostics_df['resid_add'].dropna(), lags=[6, 12], return_df=True)
print('Ljung-Box Test on Residuals:')
lb_res
```
